# Creacion modelo serie temporal

### Dividing train and test set

```python
serie_temporal.index = pd.to_datetime(serie_temporal['first_contact_date'])
serie_temporal.drop(columns=['first_contact_date'], inplace=True)
```

Es importante mencionar ya para el modelo prediccion de la serie temporal solo se tomo la data del 2018, ya que se observar pico sobre el 2017 y variaciones en la generacion de los leads de mal del doble y al tener tan solo 3 conversiones.

```python

train = serie_temporal[(serie_temporal.index.year == 2018) & ~((serie_temporal.index.month == 5) & (serie_temporal.index.day >= 21))]
test = serie_temporal[(serie_temporal.index.year == 2018) & (serie_temporal.index.month == 5) & (serie_temporal.index.day >= 21)]
```

```python
test
```

```python

train['conversions'] = train.groupby(train.index.date)['origin'].transform('count')

test['conversions'] = test.groupby(test.index.date)['origin'].transform('count')
```

```python
train.drop(columns=['mql_id',"landing_page_id","origin"], inplace=True)
test.drop(columns=['mql_id',"landing_page_id","origin"], inplace=True)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14,8))
plt.plot(train['count'], label='Train')
plt.plot(test['count'], label='Test')
plt.legend()
plt.title('Conversion 2018')
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
from keras.models import Sequential
from keras.layers import Dense, LSTM, Dropout, GRU, Bidirectional
from tensorflow.keras import layers, callbacks
```

### Feature Scaling

Queremos normalizar los datos de entrada antes de alimentar un modelo LSTM. La normalización ayuda a escalar los datos y facilita el aprendizaje del LSTM. Traduciremos la característicab entre cero y uno.

```python
sc = MinMaxScaler(feature_range=(0,1))
training_set_scaled = sc.fit_transform(train['count'].values.reshape(-1, 1))
```

```python
training_set_scaled.shape
```

```python
X_train = []
y_train = []
for i in range(10,len(training_set_scaled)):
    X_train.append(training_set_scaled[i-10:i,0])

    y_train.append(training_set_scaled[i,0])

X_train, y_train = np.array(X_train), np.array(y_train)
print(X_train)
print(y_train)
```

```python
X_train.shape
```

```python
X_train = np.reshape(X_train, (X_train.shape[0],X_train.shape[1],1))
```

```python
X_train
```

### LSTM Model

```python
model = Sequential([
    LSTM(units=50, return_sequences=True, input_shape=(X_train.shape[1],1)),
    layers.Dropout(0.2),

    layers.LSTM(units=50, return_sequences=True),
    layers.Dropout(0.2),

    layers.LSTM(units=50, return_sequences=True),
    layers.Dropout(0.2),

    layers.LSTM(units=50),
    layers.Dropout(0.2),

    layers.Dense(units=1),
])

model.compile(optimizer='adam',loss='mse', metrics=['mse', 'mae', 'mape'])
```

```python
model.summary()
```

### Fitting the model

```python
import tensorflow as tf # imports the tensorflow module and assigns it to the variable tf

with tf.device('/GPU:0'):
    # Tu código aquí, incluyendo EarlyStopping y model.fit
    early_stopping = callbacks.EarlyStopping(
        patience=6,
        min_delta=0.001,
        restore_best_weights=True,
    )

    history = model.fit(X_train,y_train,epochs=100,batch_size=24, callbacks=[early_stopping])
```

```python
early_stopping = callbacks.EarlyStopping(
    patience=6,
    min_delta=0.001,
    restore_best_weights=True,
)

history = model.fit(X_train,y_train,epochs=100,batch_size=24, callbacks=[early_stopping])
```

### Validation

Métricas:

Error cuadrático medio (ECM): diferencia cuadrática media entre los valores predichos y los valores reales.
Error absoluto medio (MAE): diferencia absoluta media entre los valores predichos y los valores reales.
Error medio porcentual absoluto (MAPE): calcula la diferencia porcentual entre los valores predichos y los valores reales y, a continuación, calcula la media de estas diferencias porcentuales.

```python
history_frame = pd.DataFrame(history.history)
```

```python
plt.plot(history_frame.loc[:, ['mse']])
plt.title('Mean squared error')
plt.show()
```

```python
plt.plot(history_frame.loc[:, ['mae']])
plt.title('Mean absolute error')
plt.show()
```

```python
plt.plot(history_frame.loc[:, ['mape']])
plt.title('Mean absolute percentage error')
plt.show()
```

### Predictions

```python
dataset_total = serie_temporal['count']
print(dataset_total)
inputs = dataset_total[len(dataset_total)-len(test['count'].values) - 10:].values
print(inputs)
inputs = inputs.reshape(-1,1)
print(inputs)
inputs  = sc.transform(inputs)
print(inputs)
```

```python
X_test = []
for i in range(10,len(inputs)):
    X_test.append(inputs[i-10:i,0])
X_test = np.array(X_test)
print(X_test)
X_test = np.reshape(X_test, (X_test.shape[0],X_test.shape[1],1))
print(X_test)
predicted_stock_price = model.predict(X_test)
predicted_stock_price = sc.inverse_transform(predicted_stock_price)
print(predicted_stock_price)
```
